```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./data/metabric.csv")

print(df.shape)

df.head()
```

```python
df.isna().sum()
```

```python
sns.ecdfplot(data=df, x="Overall Survival (Months)")

plt.xlabel("Overall Survival (Months)")
plt.ylabel("Proportion")
plt.title("ECDF of Overall Survival")
plt.show()
```

```python
sns.ecdfplot(
    data=df,
    x="Overall Survival (Months)",
    hue="Chemotherapy"
)

plt.xlabel("Overall Survival (Months)")
plt.ylabel("Proportion")
plt.title("ECDF of Overall Survival by Chemotherapy")
plt.show()
```

Based on the graph, I would predict that patients who received chemotherapy would have a shorter survival time. Larger proportion of these patients have survival times below the threshold.

We cannot tell from the graph alone, because we don't know if both groups had similar cancer conditions.
